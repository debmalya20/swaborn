##  Displot / distribution plot / numaricla plot

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('tips')
```

```python
df
```

```python
plt.subplot(1,2,1)
sns.histplot(df["total_bill"],bins=20,kde=True)

plt.subplot(1,2,2)
sns.histplot(df["tip"],bins=20,kde=True)

#this is how we make displot of any quantatice or any other column. using bims you can make the bins
#KDE stands for kernal density function
```

## JOINT PLOT


```python
sns.jointplot(x="total_bill",y="tip",data=df,kind='scatter') #this is scatter plot
```

```python
sns.jointplot(x="total_bill",y="tip",data=df,kind='kde') #kernal dencity
```

```python
sns.jointplot(x="total_bill",y="tip",data=df,kind='reg') #IMPORTENT
#this helps to join on the x exis and y axis of the numarical column
```

## pair plot

```python
sns.pairplot(df,hue='sex')#palette="rainbow") #using hex you can i dentifity any column 
# WE USE APLLET FOR DESIFINING
```

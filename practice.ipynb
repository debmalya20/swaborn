## DISTRIBUTION PLOT

```python
import matplotlib.pyplot as plt
import numpy as nup
import seaborn as sns
```

```python
df=sns.load_dataset('glue')
```

```python
df
```

```python
plt.subplot(1,2,1)
sns.histplot(df["Year"],bins=20,kde=True)
plt.subplot(1,2,2)
sns.histplot(df["Score"],bins=20,kde=True)
```

```python
sns.jointplot(x=df["Year"],y=df["Score"],kind="kde")
```

```python
sns.pairplot(df,hue="Task")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data=sns.load_dataset('tips')
data
```

```python
plt.subplot(1,2,1)
sns.histplot(data['total_bill'],bins=20,kde=True)
plt.subplot(1,2,2)
sns.histplot(data['tip'])
```

```python
sns.jointplot(x=data["total_bill"],y=data["tip"],data=data,kind="hex")
```

```python
sns.pairplot(data,hue="smoker")
```
